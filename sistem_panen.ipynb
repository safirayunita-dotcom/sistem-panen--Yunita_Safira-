```python
import os
import shutil

# Define the project directory
project_dir_name = "sistem-panen"
# Get the absolute path for robustness
project_absolute_path = os.path.join(os.getcwd(), project_dir_name)
file_path = os.path.join(project_absolute_path, "program_panen.py")

# --- Cleanup previous runs to ensure a clean state ---
if os.path.exists(project_absolute_path):
    print(f"--- Menghapus direktori proyek lama: {project_absolute_path} ---")
    shutil.rmtree(project_absolute_path)

# 1. Membuat folder proyek 'sistem-panen' dan file Python awal
print(f"--- Membuat direktori proyek baru: {project_absolute_path} ---")
!mkdir -p {project_absolute_path}

# Change Python's current working directory to the project directory
os.chdir(project_absolute_path)
print(f"--- Current working directory diubah ke: {os.getcwd()} ---")

# Menulis kode Python ke file menggunakan operasi file Python
python_code_initial = """def hitung_total_panen(jumlah, harga):
    return jumlah * harga

if __name__ == '__main__':
    jumlah = 100
    harga = 2000 # Contoh: 100 kg * Rp 2000/kg
    hasil_panen_pertama = hitung_total_panen(jumlah, harga)
    print(f'Total hasil panen: Rp {hasil_panen_pertama}')
"""
with open("program_panen.py", "w") as f: # Use relative path since we chdir'd
    f.write(python_code_initial)

print(f"\n--- Proyek '{project_dir_name}' dan file awal 'program_panen.py' berhasil dibuat ---")
print("Isi file program_panen.py:")
!cat program_panen.py
print("\nIsi direktori sistem-panen:")
!ls -R

# 2. Inisialisasi repositori Git dan commit pertama
!git init
!git add program_panen.py
!git config user.email "colab_user@example.com" # Konfigurasi email user
!git config user.name "Colab User" # Konfigurasi nama user
!git branch -M main # Mengganti nama branch awal menjadi main
!git commit -m "Initial commit: Menambahkan program penghitung total hasil panen"

print("\n--- Repositori Git diinisialisasi dan commit pertama dilakukan ---")
print("Log Git:")
!git log --oneline

# 3. Buat branch baru 'fitur-diskon', tambahkan fungsi diskon, dan commit
# Using 'git checkout -b' to create and switch, which fails if branch exists, so we check first
branch_name = "fitur-diskon"
# Check if the branch exists. os.system returns 0 for success (branch exists here).
if not os.system(f"git rev-parse --verify {branch_name} >/dev/null 2>&1") == 0:
    print(f"Creating and checking out new branch '{branch_name}'.")
    !git checkout -b {branch_name}
else:
    print(f"Branch '{branch_name}' already exists, checking it out.")
    !git checkout {branch_name}

# Untuk menghindari konflik, kita akan menulis ulang seluruh file agar mencakup fungsi diskon baru
# Ini mensimulasikan pengembangan fitur baru pada branch 'fitur-diskon'
python_code_with_discount = """def hitung_total_panen(jumlah, harga):
    return jumlah * harga

def hitung_diskon(total, persentase_diskon):
    return total * (persentase_diskon / 100)

if __name__ == '__main__':
    jumlah = 100
    harga = 2000
    hasil_panen_pertama = hitung_total_panen(jumlah, harga)
    diskon_pertama = hitung_diskon(hasil_panen_pertama, 10) # Diskon 10%
    total_setelah_diskon = hasil_panen_pertama - diskon_pertama
    print(f'Total hasil panen: Rp {hasil_panen_pertama}')
    print(f'Diskon (10%): Rp {diskon_pertama}')
    print(f'Total setelah diskon: Rp {total_setelah_diskon}')
"""
with open("program_panen.py", "w") as f:
    f.write(python_code_with_discount)

!git add program_panen.py
!git commit -m "feat: Menambahkan fungsi perhitungan diskon"

print("\n--- Branch 'fitur-diskon' dibuat, fungsi diskon ditambahkan, dan di-commit ---")
print("Log Git di branch fitur-diskon:")
!git log --oneline
print("\n--- Isi file program_panen.py di branch fitur-diskon ---")
!cat program_panen.py

# 4. Kembali ke branch utama (main) dan lakukan merge 'fitur-diskon'
!git checkout main
!git merge fitur-diskon

print("\n--- Kembali ke branch 'main' dan 'fitur-diskon' berhasil di-merge ---")
print("Log Git setelah merge:")
!git log --oneline --graph --all
print("\n--- Isi file program_panen.py setelah merge ke main ---")
!cat program_panen.py

# 5. Jelaskan perbedaan fungsi perintah 'git pull' dan 'git merge'
print("\n--- Perbedaan 'git pull' dan 'git merge' ---")
print("\n**Git Pull:**\n- Digunakan untuk mengambil (fetch) perubahan dari remote repository dan langsung menggabungkannya (merge) ke branch lokal saat ini.\n- Ini adalah kombinasi dari `git fetch` dan `git merge`.\n- Contoh: `git pull origin main` akan mengambil semua perubahan dari branch 'main' di remote 'origin' dan menggabungkannya ke branch lokal 'main' Anda.\n- Dalam konteks kerja tim, `git pull` sering digunakan untuk memperbarui branch lokal Anda dengan perubahan terbaru dari tim di remote repository.")
print("\n**Git Merge:**\n- Digunakan untuk menggabungkan dua atau lebih histori pengembangan yang berbeda menjadi satu.\n- Ini mengambil perubahan dari satu branch dan mengaplikasikannya ke branch lain.\n- Contoh: `git merge fitur-diskon` (ketika berada di branch 'main') akan menggabungkan semua perubahan dari branch 'fitur-diskon' ke branch 'main' saat ini.\n- Dalam konteks kerja tim, `git merge` adalah cara utama untuk mengintegrasikan fitur baru atau perbaikan bug dari branch pengembangan ke branch utama (seperti 'main' atau 'develop').")
print("\n**Singkatnya:**\n- `git pull` = `git fetch` + `git merge` (dari remote ke lokal)\n- `git merge` = menggabungkan dua branch lokal atau lokal dengan hasil fetch dari remote.")
```
